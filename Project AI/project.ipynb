```python
df['Saving accounts'] = df['Saving accounts'].fillna('unknown')
df['Checking account'] = df['Checking account'].fillna('unknown')
```

```python
encoders = {}
columns_to_encode = ['Sex', 'Housing', 'Saving accounts', 'Checking account', 'Purpose']

def encodeCols(cols):
    for i in cols:
        value = pd.DataFrame({i:df[i].unique()})
        LE = LabelEncoder()
        LE.fit(value)
        df[i] = LE.transform(df[i])
        encoders[i] = LE

columns = ['Sex', 'Job', 'Housing', 'Saving accounts', 'Checking account', 'Purpose']
encodeCols(columns)

# Save encoders
joblib.dump(encoders, "encoders.pkl")
```

```python
X = df.drop(['Risk', 'id'], axis=1) 
y = df['Risk']
```

```python
y = y.map({'good': 0, 'bad': 1})
```

```python
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.2, random_state=100)
```

```python
from xgboost import XGBClassifier
# Initializing the model with professional parameters
model = XGBClassifier(n_estimators=100, learning_rate=0.1, max_depth=5, random_state=100)
```

```python
model.fit(train_X, train_y)
```

```python
import joblib

joblib.dump(model,open ('credit_risk_model.pkl','wb'))
```

```python
model = joblib.load(open('credit_risk_model.pkl', 'rb'))
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(test_X)
```

```python
accuracy = accuracy_score(test_y, y_pred)
print(f"Final Model Accuracy: {accuracy * 100:.2f}%")

print("Classification Report:", end='')
print(classification_report(test_y, y_pred))
```

```python
import numpy as np

def encode_cols(le, value):
    if value in le.classes_:
        return le.transform([value])[0]
    else:
        return le.transform([le.classes_[0]])[0]

print("Enter applicant details to predict credit risk:")
applicant_data = {
    'Age': int(input("Age: ")),
    'Sex': input("Sex (male/female): ").lower(),
    'Job': int(input("Job (0-3): ")),  
    'Housing': input("Housing (own/rent/free): ").lower(),
    'Saving accounts': input("Saving accounts (little/moderate/rich/quite rich/unknown): ").lower(),
    'Checking account': input("Checking account (little/moderate/rich/quite rich/unknown): ").lower(),
    'Credit amount': int(input("Credit amount: ")),
    'Duration': int(input("Duration (months): ")),
    'Purpose': input("Purpose (car/education/furniture/electronics/repairs/other): ").lower()
}

# Encode categorical features
for col in ['Sex', 'Job', 'Housing', 'Saving accounts', 'Checking account', 'Purpose']:
    le = encoders[col]
    applicant_data[col] = encode_cols(le, applicant_data[col])

features = [
    'Age', 'Sex', 'Job', 'Housing', 'Saving accounts', 
    'Checking account', 'Credit amount', 'Duration', 'Purpose'
]

X = np.array([[applicant_data[col] for col in features]])

predict = model.predict(X)
print("Predicted Credit Risk:", "Bad" if predict[0] == 1 else "Good")
```
