```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, BernoulliNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn import metrics
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
```

```python
TOP = 20
top_genres = df['track_genre'].value_counts().head(TOP).index.tolist()
print('Genres selected for ML:', top_genres)

ml_df = df[df['track_genre'].isin(top_genres)].copy()
print('ML dataset size:', ml_df.shape)
```

```python
X_ml = ml_df[FEATURES]
y_ml = LabelEncoder().fit_transform(ml_df['track_genre'])
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X_ml, y_ml, test_size=0.3, random_state=42
)
```

```python
print('Train size:', X_train.shape)
print('Test size :', X_test.shape)
```

```python
Dtree = DecisionTreeClassifier(max_depth=10, random_state=50)
Dtree.fit(X_train, Y_train)
Y_pred = Dtree.predict(X_test) 

accuracy = accuracy_score(Y_test, Y_pred)
precision = metrics.precision_score(Y_test, Y_pred, average='weighted', zero_division=0)
```

```python
print(f'Accuracy: {accuracy:.2f}')
print(f'Precision: {precision:.2f}')    
```

```python
import pandas as pd
import numpy as np
import joblib
from sklearn.preprocessing import LabelEncoder

# ── Load model ──────────────────────────────────────────────────────────────
model = joblib.load('spotify.pkl')

# ── Rebuild the label encoder from the dataset (same genres used in training) ──
df = pd.read_csv('spotify-tracks-dataset-detailed.csv')
df.dropna(inplace=True)

TOP = 20
top_genres = df['track_genre'].value_counts().head(TOP).index.tolist()
ml_df = df[df['track_genre'].isin(top_genres)].copy()

le = LabelEncoder()
le.fit(ml_df['track_genre'])        # same encoder used during training

# ── Feature list (must match training exactly) ───────────────────────────────
FEATURES = [
    "popularity", "danceability", "energy", "key", "loudness",
    "mode", "speechiness", "acousticness", "instrumentalness",
    "liveness", "valence", "tempo", "time_signature", "explicit", "duration_min"
]

# ── Helper: safely cast float inputs to int (as done during preprocessing) ───
def to_int(val):
    try:
        return int(float(val))
    except (ValueError, TypeError):
        return 0

# ── Collect track details from user ─────────────────────────────────────────
print("Enter track details to predict its genre:\n")

track_data = {
    "popularity":        to_int(input("Popularity (0–100): ")),
    "danceability":      to_int(input("Danceability (0.0–1.0): ")),
    "energy":            to_int(input("Energy (0.0–1.0): ")),
    "key":               to_int(input("Key (0–11): ")),
    "loudness":          to_int(input("Loudness (dB, e.g. -6): ")),
    "mode":              to_int(input("Mode (0=minor, 1=major): ")),
    "speechiness":       to_int(input("Speechiness (0.0–1.0): ")),
    "acousticness":      to_int(input("Acousticness (0.0–1.0): ")),
    "instrumentalness":  to_int(input("Instrumentalness (0.0–1.0): ")),
    "liveness":          to_int(input("Liveness (0.0–1.0): ")),
    "valence":           to_int(input("Valence (0.0–1.0): ")),
    "tempo":             to_int(input("Tempo (BPM, e.g. 120): ")),
    "time_signature":    to_int(input("Time Signature (e.g. 4): ")),
    "explicit":          to_int(input("Explicit (0=No, 1=Yes): ")),
    "duration_min":      to_int(input("Duration in ms (e.g. 200000): ")),
}

# ── Predict ──────────────────────────────────────────────────────────────────
# ── Predict ──────────────────────────────────────────────────────────────────
X = np.array([[track_data[f] for f in FEATURES]])
pred_encoded = model.predict(X)[0]
pred_genre   = le.inverse_transform([pred_encoded])[0]

# ── Filter dataset by predicted genre & find closest song ────────────────────
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics.pairwise import cosine_similarity

MATCH_FEATURES = [
    "popularity", "danceability", "energy", "key", "loudness",
    "mode", "speechiness", "acousticness", "instrumentalness",
    "liveness", "valence", "tempo", "time_signature", "explicit", "duration_ms"
]

genre_df = df[df['track_genre'] == pred_genre].copy()
genre_df.reset_index(drop=True, inplace=True)

scaler = MinMaxScaler()
genre_scaled = scaler.fit_transform(genre_df[MATCH_FEATURES])

user_input = {
    "popularity": track_data["popularity"], "danceability": track_data["danceability"],
    "energy": track_data["energy"], "key": track_data["key"],
    "loudness": track_data["loudness"], "mode": track_data["mode"],
    "speechiness": track_data["speechiness"], "acousticness": track_data["acousticness"],
    "instrumentalness": track_data["instrumentalness"], "liveness": track_data["liveness"],
    "valence": track_data["valence"], "tempo": track_data["tempo"],
    "time_signature": track_data["time_signature"], "explicit": track_data["explicit"],
    "duration_ms": track_data["duration_min"]   # your notebook saved it as duration_min
}

user_vec = scaler.transform(pd.DataFrame([user_input]))
sims     = cosine_similarity(user_vec, genre_scaled)[0]
top_idx  = np.argsort(sims)[::-1][:5]

print(f"\nPredicted Genre : {pred_genre}")
print(f"\n🎵 Top 5 Recommended Songs in '{pred_genre}':\n")
for rank, i in enumerate(top_idx, start=1):
    row = genre_df.iloc[i]
    print(f"  #{rank}  {row['track_name']}  —  {row['artists']}  ({sims[i]*100:.1f}% match)")
```
